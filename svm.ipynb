```python
# 1. Feature Engineering: Let's encode the categorical features
from sklearn.preprocessing import LabelEncoder

# Make a copy of df to avoid mutating the original data
df_encoded = df.copy()

# Encode categorical variables
encoders = {}
for col in ['Geopolitical_Zone', 'Stage_at_Diagnosis', 'Histological_Type', 'Molecular_Subtype']:
    le = LabelEncoder()
    df_encoded[col] = le.fit_transform(df[col])
    encoders[col] = le

# Feature matrix 'x' (excluding ID and Target) and Target 'y'
x = df_encoded[['Age', 'Geopolitical_Zone', 'Stage_at_Diagnosis', 'Histological_Type', 'Molecular_Subtype']]
y = df_encoded['Primary_Treatment_Plan']

# Train-test split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
# 2. Standardize numerical features using standard variable naming ('scaler')
scaler = StandardScaler()
x_train = scaler.fit_transform(x_train)
x_test = scaler.transform(x_test)
```

```python
# 3. Fit and Predict using correctly initialized and named SVC model
print("=====RBF SVM (NON-LINEAR) RECOMMENDED=====")
rbf_svm = SVC(kernel="rbf", C=1.0, gamma='scale', random_state=42)
rbf_svm.fit(x_train, y_train)
pred_rbf = rbf_svm.predict(x_test)

print(f"accuracy: {accuracy_score(y_test, pred_rbf)*100:.2f}%")
print(classification_report(y_test, pred_rbf))
print("confusion_matrix:\n", confusion_matrix(y_test, pred_rbf))
```
